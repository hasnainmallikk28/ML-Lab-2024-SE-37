## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).
- **(g)** Monsoon (Jul–Sep) rain of each city by slicing columns.
- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# T7: Climate of five cities (real data, PMD 1991–2020 normals)
print("T7a (Shape of tmax):", tmax.shape)

t7b = tmax.mean(axis=1)
print("T7b (Annual mean max temperature of each city):", t7b)

hottest_months = months[tmax.argmax(axis=1)]
print("T7c (Hottest month of each city):", hottest_months)

t7d = tmax.mean(axis=0)
print("T7d (5-city average for each month):", t7d)

t7e = (tmax > 35).sum(axis=1)
print("T7e (Months above 35 °C in each city):", t7e)

t7f = tmax[0] * 9/5 + 32
print("T7f (Lahore's temperatures in °F):", t7f)

# Monsoon is Jul to Sep (indices 6 to 8)
monsoon_rain = rain[:, 6:9].sum(axis=1)
print("T7g (Monsoon rain Jul-Sep per city):", monsoon_rain)

wettest_monsoon_city = cities[monsoon_rain.argmax()]
yearly_rain = rain.sum(axis=1)
monsoon_share = (monsoon_rain / yearly_rain) * 100
print("T7h (Wettest city in monsoon):", wettest_monsoon_city)
print("T7h (% of yearly rain in monsoon for each city):", monsoon_share)
```

#### Interpretation:
#####     Islamabad recording the highest monsoon rainfall (over 760 mm) rather than Lahore was surprising, demonstrating how sub-mountainous regions receive significantly heavier downpours during the monsoon season.

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
`cpi` has 2 rows (FY2024, FY2025) and 10 columns (Jul to Apr), year-on-year CPI inflation in %.
- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.
- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# T8: Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
t8a = cpi.mean(axis=1)
print("T8a (Average inflation of each year):", t8a)

t8b = cpi_months[cpi[0] > 25]
print("T8b (FY2024 months with inflation above 25%):", t8b)

t8c = cpi[0] - cpi[1]
print("T8c (Drop in inflation for each month FY24 - FY25):", t8c)

biggest_drop_month = cpi_months[t8c.argmax()]
print("T8d (Month with the biggest drop):", biggest_drop_month)

t8e = np.where(cpi[1] < 5, "Low", np.where(cpi[1] <= 10, "Moderate", "High"))
print("T8e (FY2025 inflation labels):", t8e)
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9: Grace marks
marks_grace = np.clip(marks + 5, 0, 100)
print("T9a (Marks after 5 grace marks, capped at 100):", marks_grace)

pass_percentage = (np.sum(marks_grace >= 50) / len(marks_grace)) * 100
print("T9b (Pass % after grace):", pass_percentage)

passed_only_grace = np.where((marks < 50) & (marks_grace >= 50))[0]
print("T9c (Indices of students who passed only because of grace marks):", passed_only_grace)

z_scores = (marks_grace - marks_grace.mean()) / marks_grace.std()
print("T9d (Z-scores after grace):", z_scores)
```
