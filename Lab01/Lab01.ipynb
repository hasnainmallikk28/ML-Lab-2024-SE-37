### Task 06: Tuple vs List Difference

**Problem Statement:** What is the main difference between a tuple and a list?  
**Feature Used:** Data Structures Theory (Mutability vs. Immutability)



```python
#  difference between a list and a tuple
# Lists use square brackets [] and are MUTABLE (can be changed or modified).
# Tuples use parentheses () and are IMMUTABLE (cannot be changed or modified).

print("1. List: Mutable (Values can be modified, added, or removed after creation).")
print("2. Tuple: Immutable (Values are fixed and cannot be changed once created).")
```

### Task 07: Email Domain Extractor Function

**Problem Statement:** Write a function that grabs the email website domain from a string in the form: `user@domain.com`.  
**Feature Used:** User-Defined Functions & String Manipulation (`split()` method)



```python
email= "user@domain.com" 
def getDomain(email):
    parts=email.split('@')
    return parts[1]
print(getDomain(email))
```

### Task 08: Substring Finder Function

**Problem Statement:** Write a function that returns `True` if the word `"dog"` is contained in the input string (handle case-insensitivity).  
**Feature Used:** Functions, Conditionals, & Case-Insensitive String Checks (`in` operator)



```python
def find_dog(input_string):
    lower_text = input_string.lower()
    if "dog" in lower_text:
        return True
    else:
        return False
string1 = "Look there is a dog"
string2 = "Cat and dog are running"
string3 = "Parrot is sitting in a cage"

print("Test 1 :", find_dog(string1))
print("Test 2 :", find_dog(string2))
print("Test 3 :", find_dog(string3))
```

### Task 09: Word Counter Function

**Problem Statement:** Write a function that counts the number of times the word `"dog"` occurs in a string (handle case-insensitivity).  
**Feature Used:** Functions & String Counting Method (`count()`)



```python
def count_dog(input_string):
    lower_text = input_string.lower()
    return lower_text.count('dog')
string1 = "This dog runs faster than that DOG and doggy."
string2 = "No dogs allowed here at all."
print("Test 1 :", count_dog(string1))
print("Test 2 :", count_dog(string2))
```

### Task 10: Speeding Ticket Logic

**Problem Statement:** Write a function `caught_speeding(speed, is_birthday)` with custom boundary rules and birthday allowance (+5 mph).
* Speed $\le$ 60: "No Ticket"
* Speed 61–80: "Small Ticket"
* Speed $\ge$ 81: "Big Ticket"
* *Condition:* On your birthday, your speed can be 5 mph higher in all cases!  
**Feature Used:** Functions, Conditional Logic (`if-elif-else`), & Arithmetic Adjustments



```python
def caught_speeding(speed,is_birthday):
    if is_birthday:
        speed -=5
        
    if speed<=60:
        return "No Ticket!"
    elif 61<= speed<=80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print(caught_speeding(81,True))
print(caught_speeding(81,False))
```
